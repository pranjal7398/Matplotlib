```python
#plotting a data on x and y axis

import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,30,40]

plt.plot(x,y)
plt.show()
```

```python
#plotting weekly sales of any company as a line graph

import matplotlib.pyplot as plt

x = ['Mon' , 'Tue' , 'Wed' , 'Thu' , 'Fri' , 'Sat']
y = [10,7,11,8,6,14]

plt.title("Weekly Sales")
plt.xlabel("Days of the week")
plt.ylabel("Sales")
plt.plot(x, y, linestyle='--', marker='o')
plt.grid(color='gray' , linestyle=':')
plt.show()
```

```python
#plotting a bar graph between products and sales

import matplotlib.pyplot as plt

products = ['p1' , 'p2' , 'p3' , 'p4' , 'p5' , 'p6']
sales = [10,7,11,8,6,4]

plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.bar(products , sales , color='green' , label ="Sales 2026")
plt.legend(loc="upper right")
plt.show()
```

```python
# plotting a pie chart of revenue collected by each region

import matplotlib.pyplot as plt

region = ["East" , "West" , "North" , "South"]
revenue = [2000 , 3500 , 1000 , 1500 ]

plt.pie(revenue , labels = region , colors= ['red' , 'skyblue' , 'orange' , 'yellow'] , autopct = "%1.1f%%")
plt.title("Regionwise Revenue")
plt.show()
```

```python
# plotting a scatter plot to visualize no. of hours studied to marks scored

import matplotlib.pyplot as plt

hours = [
    2, 4, 5, 6, 3,
    7, 8, 4, 9, 6,
    3, 5, 7, 8, 10,
    4, 6, 5, 9, 7
]

scores = [
    45, 67, 78, 89, 56,
    72, 91, 63, 84, 77,
    52, 69, 88, 74, 95,
    61, 83, 70, 58, 79
]

plt.scatter(hours , scores , color="red" , marker="o" , label="Scores")
plt.title("Student data")
plt.legend()
plt.xlabel("Hours studied")
plt.ylabel("Scores")
plt.grid(linestyle=":")
plt.show()

```

```python
# plotting a scatter plot between two students of the same class
import matplotlib.pyplot as plt

hours = [
    1,2,3,4,5
]

scores1 = [
    45,65,75,78,85
]
scores2 = [
    50,70,72,75,90
]

plt.scatter(hours , scores1 , color="orange" , marker="o" , label="Scores of student1")
plt.scatter(hours , scores2 , color="blue" , marker="o" , label="Scores of student2")
plt.legend()
plt.title("Student data")
plt.xlabel("Hours studied")
plt.ylabel("Scores")
plt.grid(linestyle=":")
plt.show()

```

```python
# subplots
# syntax :- plt.subplot( no. of rows , no. of col , index)

import matplotlib.pyplot as plt

x = [1,2,3,4]
y=[10,20,15,35]

plt.subplot(1,2,1)
plt.plot(x,y)

plt.subplot(1,2,2)
plt.bar(x,y)

plt.show()
```

```python
# subplots in case to expand the column
# syntax :- plt.subplot( no. of rows , no. of col , index)

import matplotlib.pyplot as plt

x = [1,2,3,4]
y=[10,20,15,35]

plt.subplot(2,2,1)
plt.plot(x,y)
plt.xlabel("hours studied")
plt.ylabel("scores")

plt.subplot(2,2,3)
plt.pie(y,colors=["blue" , "green" , "red" , "yellow"] , autopct = "%1.1f%%")

plt.subplot(2,2,(2,4))
plt.bar(x,y)
plt.xlabel("hours studied")
plt.ylabel("scores")

plt.show()
```

```python
#another syntax for subplots ==> fig, ax = plt.subplots( rows , col , figsize = (width , height))

import matplotlib.pyplot as plt

fig , ax = plt.subplots( 1 ,2 , figsize = (10,5)) # fig size is optional
x = [1,2,3,4]
y=[10,20,15,35]

ax[0].plot(x,y)
ax[0].set_title("line graph")

ax[1].bar(x,y)
ax[1].set_title("Bar graph")

fig.suptitle("Comparison of line and bar graph")
plt.tight_layout() # for covering full canvas
plt.show()
```

```python
# project - visualizing netflix data and plotting various graphs and charts
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2021/2021-04-20/netflix_titles.csv"
df = pd.read_csv(url)

#clean data
df = df.dropna(subset = ['type' , 'release_year' , 'rating' , 'country' , 'duration'])

typee = df['type'].value_counts()
x=typee.index
y=typee.values
plt.bar(x , y , color = ['blue' , 'orange'])
plt.title("No. of movies vs TV shows")
plt.tight_layout()
plt.show()

ratings = df['rating'].value_counts()
m = ratings.values
n = ratings.index
plt.pie(m ,labels = n , autopct = "%1.1f%%")
plt.title("Content rating")
plt.tight_layout()
plt.show()

movie_df = df[df['type'] == 'Movie'].copy()
movie_df['duration_int']  =movie_df['duration'].str.replace('min' , '').astype(int)
#movie_df['duration'].str.replace('min' , '') replaces min to empty string and .astype(int) converts the string values into integer values
plt.hist(movie_df['duration_int'] , bins = 30 , color ="pink" , edgecolor = "black")
plt.title("Movie duration distribution")
plt.show()

release = df['release_year'].value_counts().sort_index()
plt.scatter(release.index , release.values , color = "red")
plt.title("Release year vs No. of shows")
plt.show()

country = df['country'].value_counts().head(10)
plt.barh(country.index , country.values , color = 'teal')
plt.title("Top 10 countries by no. of shows")
plt.show()

content = df.groupby(['release_year' , 'type']).size().unstack().fillna(0)
fig , ax = plt.subplots(1,2,figsize=(12,5))
#Movies
ax[0].plot(content.index , content['Movie'] , color="blue")
ax[0].set_title('Movies released per year')
#Tv shows
ax[0].plot(content.index , content['TV Show'] , color="orange")
ax[0].set_title('TV Shows released per year')
fig.suptitle("Comparison of Movies and TV Shows")
plt.tight_layout()
plt.show()
```
