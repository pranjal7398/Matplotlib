```python
#plotting a data on x and y axis

import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,30,40]

plt.plot(x,y)
plt.show()
```

```python
#plotting weekly sales of any company as a line graph

import matplotlib.pyplot as plt

x = ['Mon' , 'Tue' , 'Wed' , 'Thu' , 'Fri' , 'Sat']
y = [10,7,11,8,6,14]

plt.title("Weekly Sales")
plt.xlabel("Days of the week")
plt.ylabel("Sales")
plt.plot(x, y, linestyle='--', marker='o')
plt.grid(color='gray' , linestyle=':')
plt.show()
```

```python
#plotting a bar graph between products and sales

import matplotlib.pyplot as plt

products = ['p1' , 'p2' , 'p3' , 'p4' , 'p5' , 'p6']
sales = [10,7,11,8,6,4]

plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.bar(products , sales , color='green' , label ="Sales 2026")
plt.legend(loc="upper right")
plt.show()
```

```python
# plotting a pie chart of revenue collected by each region

import matplotlib.pyplot as plt

region = ["East" , "West" , "North" , "South"]
revenue = [2000 , 3500 , 1000 , 1500 ]

plt.pie(revenue , labels = region , colors= ['red' , 'skyblue' , 'orange' , 'yellow'] , autopct = "%1.1f%%")
plt.title("Regionwise Revenue")
plt.show()
```

```python
# plotting a scatter plot to visualize no. of hours studied to marks scored

import matplotlib.pyplot as plt

hours = [
    2, 4, 5, 6, 3,
    7, 8, 4, 9, 6,
    3, 5, 7, 8, 10,
    4, 6, 5, 9, 7
]

scores = [
    45, 67, 78, 89, 56,
    72, 91, 63, 84, 77,
    52, 69, 88, 74, 95,
    61, 83, 70, 58, 79
]

plt.scatter(hours , scores , color="red" , marker="o" , label="Scores")
plt.title("Student data")
plt.legend()
plt.xlabel("Hours studied")
plt.ylabel("Scores")
plt.grid(linestyle=":")
plt.show()

```

```python
# plotting a scatter plot between two students of the same class
import matplotlib.pyplot as plt

hours = [
    1,2,3,4,5
]

scores1 = [
    45,65,75,78,85
]
scores2 = [
    50,70,72,75,90
]

plt.scatter(hours , scores1 , color="orange" , marker="o" , label="Scores of student1")
plt.scatter(hours , scores2 , color="blue" , marker="o" , label="Scores of student2")
plt.legend()
plt.title("Student data")
plt.xlabel("Hours studied")
plt.ylabel("Scores")
plt.grid(linestyle=":")
plt.show()

```

```python
# subplots
# syntax :- plt.subplot( no. of rows , no. of col , index)

import matplotlib.pyplot as plt

x = [1,2,3,4]
y=[10,20,15,35]

plt.subplot(1,2,1)
plt.plot(x,y)

plt.subplot(1,2,2)
plt.bar(x,y)

plt.show()
```

```python
# subplots in case to expand the column
# syntax :- plt.subplot( no. of rows , no. of col , index)

import matplotlib.pyplot as plt

x = [1,2,3,4]
y=[10,20,15,35]

plt.subplot(2,2,1)
plt.plot(x,y)
plt.xlabel("hours studied")
plt.ylabel("scores")

plt.subplot(2,2,3)
plt.pie(y,colors=["blue" , "green" , "red" , "yellow"] , autopct = "%1.1f%%")

plt.subplot(2,2,(2,4))
plt.bar(x,y)
plt.xlabel("hours studied")
plt.ylabel("scores")

plt.show()
```
